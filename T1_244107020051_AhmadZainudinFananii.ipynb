### 3. Kegunaan Masing-Masing Library

* **a. PyPREP**  
  Digunakan untuk melakukan pemrosesan otomatis pada sinyal gelombang otak atau **EEG** (*Electroencephalogram*).  
  * *Sumber:* [PyPREP Documentation](https://pyprep.readthedocs.io/)

* **b. Scipy**  
  Digunakan untuk fungsi-fungsi matematika tingkat lanjut, optimasi, aljabar linear, statistik, serta penerapan filter digital (seperti *butterworth filter*) pada sinyal.  
  * *Sumber:* [SciPy Documentation](https://scipy.org/)

* **c. wandb**  
  Digunakan untuk membantu mencatat (*logging*) metrik pelatihan model secara *real-time*, memvisualisasikan grafik performa, melacak *hyperparameters*, serta menyimpan versi model dan dataset agar mudah dievaluasi.  
  * *Sumber:* [Weights & Biases (wandb) Documentation](https://docs.wandb.ai/)

* **d. pyECG**  
  Digunakan untuk memproses, menganalisis, dan mengekstrak fitur dari sinyal aktivitas listrik jantung atau **EKG/ECG** (*Electrocardiogram*). Library ini membantu dalam analisis detak jantung dan pengolahan data rekam medis jantung.  
  * *Sumber:* [pyECG di PyPI](https://pypi.org/project/pyECG/)
